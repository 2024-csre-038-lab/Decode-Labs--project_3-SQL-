```python
# ==============================================================================
# DECODELABS: DATA ANALYTICS PROJECT 3 - SQL DATA ANALYSIS
# Industrial Training Kit | Batch: 2026
# ==============================================================================

import sqlite3
import pandas as pd

print("Connecting to the SQL Database Engine...")
# 1. Database se connect karna (Yeh system memory me ek temporary SQL DB bana dega)
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# ------------------------------------------------------------------------------
# STEP 1: DATABASE SETUP & DATA INGESTION
# ------------------------------------------------------------------------------
print("\n[Step 1] Creating 'EmployeeSales' Table and inserting raw data...")

# Table create karne ki SQL Query
cursor.execute('''
CREATE TABLE EmployeeSales (
    EmployeeID INT PRIMARY KEY,
    EmployeeName VARCHAR(50),
    Role VARCHAR(50),
    Salary INT,
    TotalSales INT,
    Email VARCHAR(100),
    Month VARCHAR(20)
);
''')

# Raw Data insert karne ki SQL Query (As per Project requirements)
raw_data = [
    (101, 'Amit Sharma', 'Developer', 55000, 1200, 'amit@gmail.com', 'January'),
    (102, 'Sana Khan', 'Marketer', 45000, 850, 'sana@gmail.com', 'January'),
    (103, 'John Doe', 'Developer', 60000, 1500, 'john@yahoo.com', 'February'),
    (104, 'Rahul Verma', 'Designer', 48000, 900, 'rahul@gmail.com', 'January'),
    (105, 'Zainab Bi', 'Marketer', 47000, 1100, 'zainab@gmail.com', 'February'),
    (106, 'Vikram Singh', 'Designer', 52000, 0, 'vikram@yahoo.com', 'February')
]

cursor.executemany("INSERT INTO EmployeeSales VALUES (?, ?, ?, ?, ?, ?, ?);", raw_data)
conn.commit()
print("Database Setup Completed Successfully!")

# ------------------------------------------------------------------------------
# STEP 2: EXECUTING SQL QUERIES & RESULT SETS
# ------------------------------------------------------------------------------
print("\n" + "="*60)
print("STEP 2: EXECUTING ANALYTICAL SQL QUERIES")
print("="*60)

# --- Query 1: Basic Selection, Row Filtering & Sorting ---
# Requirement: Filter employees with Salary >= 50000 and using Gmail, sorted by Salary desc.
print("\n>>> QUERY 1: High Earners using Gmail (WHERE & ORDER BY) <<<")
query_1 = """
SELECT EmployeeID, EmployeeName, Role, Salary, Email 
FROM EmployeeSales
WHERE Salary >= 50000 AND Email LIKE '%@gmail.com'
ORDER BY Salary DESC;
"""
df_res1 = pd.read_sql_query(query_1, conn)
print(df_res1.to_string(index=False))


# --- Query 2: Categorical Bucketing & Aggregations ---
# Requirement: Group by Role to find count, average salary, and sum of sales.
print("\n>>> QUERY 2: Role-wise Structural Breakdown (GROUP BY & AGGREGATIONS) <<<")
query_2 = """
SELECT 
    Role, 
    COUNT(*) AS TotalEmployees, 
    AVG(Salary) AS AverageSalary, 
    SUM(TotalSales) AS CombinedSales
FROM EmployeeSales
GROUP BY Role
ORDER BY CombinedSales DESC;
"""
df_res2 = pd.read_sql_query(query_2, conn)
print(df_res2.to_string(index=False))


# --- Query 3: Advanced Filtering on Aggregated Groups ---
# Requirement: Filter roles where combined sales are greater than 1000.
print("\n>>> QUERY 3: High Performing Roles (HAVING Clause) <<<")
query_3 = """
SELECT Role, SUM(TotalSales) AS NetSales
FROM EmployeeSales
GROUP BY Role
HAVING SUM(TotalSales) > 1000;
"""
df_res3 = pd.read_sql_query(query_3, conn)
print(df_res3.to_string(index=False))

# Database connection close karna
conn.close()

# ------------------------------------------------------------------------------
# STEP 3: LOGICAL FORENSIC INSIGHTS & THEORETICAL FRAMEWORK
# ------------------------------------------------------------------------------
print("\n" + "="*60)
print("STEP 3: LOGICAL SQL ARCHITECTURE & FORENSIC INSIGHTS")
print("="*60)
print("""
1. DATA COMPLIANCE & EXECUTION ORDER:
   SQL syntax me SELECT sabse upar likha jata hai, lekin Database Engine ise sabse aakhir me 
   process karta hai. Is project me engine ka actual execution flow ye raha:
   FROM -> WHERE -> GROUP BY -> HAVING -> SELECT -> ORDER BY.

2. AVOIDING THE ALIAS TRAP:
   Query 3 me filter lagane ke liye WHERE clause ka istemal nahi kiya ja sakta tha, kyunki 
   WHERE clause GROUP BY se pehle chalta hai aur us waqt tak 'NetSales' (Alias) banta hi nahi hai. 
   Isliye humne groups ko filter karne ke liye 'HAVING SUM(TotalSales) > 1000' ka sahi istemal kiya.

3. NULL & ZERO VALUE INTEGRITY:
   Vikram Singh (ID 106) ki sales '0' thi. COUNT(*) pure records ko ginta hai jabki SUM aur AVG 
   sirf numeric math apply karte hain. Hamare database me '0' sales hone ke bawajud data complete 
   hai aur structural integrity barkarar hai.
""")
```
